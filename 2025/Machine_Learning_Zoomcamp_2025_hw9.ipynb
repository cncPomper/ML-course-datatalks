```python
PREFIX="https://github.com/alexeygrigorev/large-datasets/releases/download/hairstyle"
DATA_URL=f"{PREFIX}/hair_classifier_v1.onnx.data"
MODEL_URL=f"{PREFIX}/hair_classifier_v1.onnx"
!wget {DATA_URL}
!wget {MODEL_URL}
```

```python
import onnxruntime as rt

sess = rt.InferenceSession("hair_classifier_v1.onnx", providers=rt.get_available_providers())

# Get output node names
output_names = [output.name for output in sess.get_outputs()]

print(f"The output node name is: {output_names[0]}")
```
